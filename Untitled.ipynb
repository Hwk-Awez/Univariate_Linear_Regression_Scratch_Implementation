```python
x=np.array(df['YearsExperience'])
y=np.array(df['Salary'])
```

```python
plt.scatter(x,y)
plt.show()
```

```python
#No. of datapoints

m=x.size
print("Size : ",m)
```

```python
#Initialise slope,intercept,epochs and learning rate
theta=0
c=0

epochs=100
alpha=0.001
```

```python
def cost_func(theta,c,m,x,y):

    sum=0

    for i in range(m):
        sum += ((theta * x[i] + c) - y[i])**2

    return (sum / (2 * m))
```

```python
cost_func(theta,c,m,x,y)
```

```python
grad_mse=[cost_func(theta,c,m,x,y)]
grad_theta=[0]
grad_int=[0]
```

```python
for i in range(epochs):

    cost_rate=0
    int_rate=0

    for j in range(m):

        cost_rate += x[j] * ((theta * x[j]) + c - y[j])

    cost_rate /= m

    for k in range(m):

        int_rate += (theta * x[k]) + c - y[k]

    int_rate /= m

    theta = theta - (alpha * cost_rate)
    c = c - (alpha * int_rate)

    grad_mse.append(cost_func(theta,c,m,x,y))
    grad_theta.append(theta)
    grad_int.append(c)

    print(cost_func(theta,c,m,x,y))

print("Slope : ",theta)
print("Intercept : ",c)    
```

```python
plt.scatter(x,y)
plt.plot(x, (theta * x) + c)
plt.show()
```

```python
grad_mse=np.array(grad_mse)
grad_theta=np.array(grad_theta)
grad_int=np.array(grad_int)
```

```python
plt.scatter(grad_theta,grad_mse)
plt.show()
```
